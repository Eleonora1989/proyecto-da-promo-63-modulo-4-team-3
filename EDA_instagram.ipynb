```python
import pandas as pd
 
# cargar dataset
df_instagram = pd.read_csv("data/instagram_usage_lifestyle.csv")
 
# seleccionar columnas
columns = [
    "user_id","age","gender","country","income_level", "education_level", "employment_status", "daily_active_minutes_instagram","posts_created_per_week",
    "likes_given_per_day","comments_written_per_day","followers_count",
    "perceived_stress_score","self_reported_happiness","sleep_hours_per_night",
    "exercise_hours_per_week", "daily_steps_count", "weekly_work_hours",
    "relationship_status", "time_on_feed_per_day", "time_on_reels_per_day", "smoking", "alcohol_frequency", "diet_quality", "posts_created_per_week", "comments_written_per_day", "dms_sent_per_week"
]

df_instagram = df_instagram[columns]

columnas_float = ["daily_active_minutes_instagram", "sleep_hours_per_night", "exercise_hours_per_week", "weekly_work_hours"]
 
 
for col in columnas_float:
    df_instagram[col] = df_instagram[col].astype(int)
 
 
df_instagram.head(1)
 
# muestreo (50%)
df_instagram = df_instagram.sample(frac=0.5, random_state=42)
 
# guardar nuevo archivo
df_instagram.to_csv("clean_data/instagram_reducido.csv", index=False)
```
