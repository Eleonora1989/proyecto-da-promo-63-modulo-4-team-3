```python
df_instagram = pd.read_csv("data/instagram_usage_lifestyle.csv")
```

```python
def exploracion_inicial(df,name="dataframe"):
    print(f"El dataframe '{name}' tiene {df.shape[0]} filas y {df.shape[1]} columnas.")
    display(df.head(2))
    print("="*150)
    display(df.tail(2))
    print("="*150)
    display(df.sample(2))
    print("="*150)
    print(f"Listamos abajo el conteo de no-nulos junto con el tipo de dato:")
    print("="*150)
    print(df.info())
    print("="*150)
```

```python
exploracion_inicial(df_instagram, "df_instagram")
```

```python
def descriptive_analysis(df):
    print("NUMÉRICAS")
    display(df.describe(include="number").T)
    print("CATEGÓRICAS")
    display(df.describe(include="object").T)
```

```python
descriptive_analysis(df_instagram)
```

```python
def analizar_categorias(df):
    col_cat = df.select_dtypes(include = "object").columns

    for col in col_cat:
        print("=" * 100)
        print(f"Analizamos las categorias de la columna: {col}")
        print("=" * 100)
        print(df[col].unique())
```

```python
analizar_categorias(df_instagram)
```

```python
def preparar_datos_instagram(df):
    """
    Filtra las columnas relevantes para el análisis de
    perfil de usuario y salud mental en Instagram.
    """
    
    columnas_a_conservar = [
        # Demográficas
        'age',
        'gender',
        'country',
        'urban_rural',
        'income_level',
        'education_level',
        'employment_status',
        
        # Uso de la app
        'daily_active_minutes_instagram',
        'sessions_per_day',
        'time_on_reels_per_day',
        'time_on_feed_per_day',
        'time_on_explore_per_day',
        'content_type_preference',
        'preferred_content_theme',
        'user_engagement_score',
        
        # Bienestar y salud mental
        'perceived_stress_score',
        'self_reported_happiness',
        'sleep_hours_per_night',
        'exercise_hours_per_week',
        'smoking',
        'alcohol_frequency',
        'body_mass_index',
    ]
    
    # Verificar que todas las columnas existen
    columnas_faltantes = [col for col in columnas_a_conservar if col not in df.columns]
    if columnas_faltantes:
        print(f"⚠️ Columnas no encontradas: {columnas_faltantes}")
    
    columnas_validas = [col for col in columnas_a_conservar if col in df.columns]
    df_filtrado = df[columnas_validas].copy()
    
    print(f"✅ Columnas originales:  {df.shape[1]}")
    print(f"✅ Columnas conservadas: {df_filtrado.shape[1]}")
    print(f"✅ Filas:                {df_filtrado.shape[0]:,}")
    
    return df_filtrado


# Uso
df_tableau = preparar_datos_instagram(df_instagram)

# Exportar a CSV
df_tableau.to_csv('instagram_tableau.csv', index=False)
print("✅ CSV exportado correctamente")
```
